```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
ticker = 'BUFF'

df = yf.download(tickers = ticker, start = '2020-01-01',interval = '1d' )
df
```

```python
plt.figure(figsize=(12, 8))
plt.plot(df.index, (df['Close']))
plt.title('Portfolio Cumulative Returns')
plt.xlabel('Date')
plt.ylabel('Cumsum')

plt.show()
```

```python
df['Returns'] = np.log(df['Close']) - np.log(df['Close']).shift(1)
df
```

```python
df2 = yf.download(tickers = ticker, start = '2020-01-01',interval = '3mo' )
```

```python
df2 = df2[df2.index.month == 1]
df2['Returns'] = np.log(df2['Close']) - np.log(df2['Close']).shift(1)
df2 = df2.iloc[1:]
mean_returns = df2['Returns'].mean()
print(rf'Mean Returns : {round(mean_returns * 100, 2)} %')
df2
```
